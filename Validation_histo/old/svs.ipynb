```python
import openslide
from PIL import Image
import matplotlib.pyplot as plt
```

# Select file and patient

```python
dir="./Patient_1"
slide=openslide.OpenSlide(f"{dir}/CON_01-p53 B6.svs")

print(f"Dimensions : {slide.dimensions}")
print(f"Levels : {slide.level_count}")
print(f"Goal : {slide.properties.get('openslide.objective-power', 'Non specified')}")
```

```python
thumbnail = slide.get_thumbnail((800, 600))
plt.imshow(thumbnail)
plt.axis('off')
plt.show()
```
