# Libraries

```python
from scipy.io import loadmat
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, IntSlider
```

# Select patient and file

```python
dir="./Patient_1"
data=loadmat(f"{dir}/tumormask_qb32_D19.mat")
```

# First investigation

```python
print(data.keys())
print(data["__header__"])
print(data["__version__"])
print(data["__globals__"])
```

```python
name=list(data)[-1]
mask=data[name] #sagitarial plan
mask_t=np.transpose(mask, (0,2,1)) #transverse plan

print(f"Array shape is {mask_t.shape}")
print(f"The max of signal values is {np.max(mask_t)} at voxel {np.where(mask_t==np.max(mask_t))}")
print(f"Number of non-zero values is : {np.count_nonzero(mask_t)}")
```

# Plot a slice

```python
i_slice=35
slice=mask_t[:,:,i_slice]

plt.imshow(slice, cmap='gray')
plt.colorbar()
plt.title(f"Slice {i_slice}")
plt.axis('off')
plt.show()
```
